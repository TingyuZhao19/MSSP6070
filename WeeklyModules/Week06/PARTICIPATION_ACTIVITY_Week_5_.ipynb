```python
import pandas as pd
import os

BASE = '/content/drive/MyDrive/MSSP607/Modules/Week05'

# Read the CSV files into DataFrames
aapl = pd.read_csv(os.path.join(BASE, 'aapl.csv'))
amzn = pd.read_csv(os.path.join(BASE, 'amzn.csv'))
nflx = pd.read_csv(os.path.join(BASE, 'nflx.csv'))
fb   = pd.read_csv(os.path.join(BASE, 'fb.csv'))
goog = pd.read_csv(os.path.join(BASE, 'goog.csv'))

# Add a 'ticker' column to each DataFrame
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
nflx['ticker'] = 'NFLX'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'

# Concatenate into a single DataFrame
faanf = pd.concat([aapl, amzn, nflx, fb, goog], ignore_index=True)

display(faanf)
```

```python
# Save as 'faanf.csv' and also as 'fanng.csv' for later step
faanf_path = os.path.join(BASE, 'faanf.csv')
fanng_path = os.path.join(BASE, 'fanng.csv')
faanf.to_csv(faanf_path, index=False)
faanf.to_csv(fanng_path, index=False)

print('Saved:', faanf_path)
print('Saved:', fanng_path)
faanf.head()
```

```python
df = pd.read_csv(os.path.join(BASE, 'fanng.csv'))

df['date'] = pd.to_datetime(df['date'], errors='coerce')                # date to datetime
df['volume'] = pd.to_numeric(df['volume'], errors='coerce').fillna(0).astype('int64')  # volume to int
faang_df_sorted = df.sort_values(['date', 'ticker']).reset_index(drop=True)

faang_df_sorted.to_csv(os.path.join(BASE, 'fanng_typed_sorted.csv'), index=False)
display(faang_df_sorted.head())
```

```python
top7 = faang_df_sorted.nlargest(7, 'volume')
top7.to_csv(os.path.join(BASE, 'fanng_top7_volume.csv'), index=False)
display(top7)
```

```python
faang_melted = faang_df_sorted.melt(id_vars=['date', 'ticker'],
                                    var_name='variable', value_name='value')
faang_melted.to_csv(os.path.join(BASE, 'fanng_melted.csv'), index=False)
display(faang_melted)
```
